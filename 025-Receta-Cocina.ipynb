```python
#Programa: Receta de cocina
#|-------------Datos a ingresar----------------|
receta = input("Nombre de la receta")
ingredientes = input("Ingrese los nombres de los ingredientes")
tiempo_preparacion = int(input("Tiempo de preparacion (minutos)"))
dificultad = input("Dificultad (Facil, medo, dificil)")
print(" ")
#|-----------------Impresion---------------------|
print("--------------Receta de cocina--------------")
print(f'Nombre de la receta: {receta}')
print(f'Ingredientes a utilizar: {ingredientes}')
print(f'tiempo de preparacion: {tiempo_preparacion}')
print(f'Dificultad: {dificultad}')
```
