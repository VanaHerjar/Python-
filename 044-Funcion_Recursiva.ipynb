```python
def funcion_recursiva(numero):
    if numero == 1:
        print(numero, end= ' ' )  #1
    else:
        print(numero, end= ' ' )
        funcion_recursiva(numero - 1)
        
funcion_recursiva(8)
```

```python
def factorial_de_un_numero(numero):
    if numero == 0:
        return 1
    else:
        return numero * factorial_de_un_numero(numero - 1)

#Programa principal

factorial_de_un_numero(3)
```

```python
def potencia(base, exponente):
    if exponente == 0: 
        return 1
    else:
        return base * potencia(base, exponente - 1)

potencia(3,3)
```

```python
#Sistema de inventario

#Inventario del almacen 

inventario = [
    {'id':1, 'nombre' : 'camisa', 'precio' : 25.99, 'cantidad' : 50},
    {'id':2, 'nombre' : 'pantalones', 'precio' : 39.99, 'cantidad' : 30},
    {'id':3, 'nombre' : 'zapatos', 'precio' : 49.99, 'cantidad' : 20}
]

def mostrar_inventario():
    for producto in inventario:
        print(f'\nId: {producto.get("id")}, Nombre: {producto.get('nombre').capitalize()}, '
             f'Precio: {producto.get("precio")}, Cantidad: {producto.get("cantidad")}')

def agregar_producto():
    Id_producto = int(input(f"Ingrese el id del producto a agregar (debe ser mayor a {len(inventario)})"))
    Nombre = input("Ingrese el nombre del producto a agregar")
    Precio = input("Ingrese el precio del producto a agregar")
    Cantidad = int(input("Ingrese la cantidad del producto a agregar"))
    producto_nuevo = {'id': Id_producto, 'nombre' : Nombre.capitalize(), 'precio' : Precio, 'cantidad' : Cantidad}
    inventario.append(producto_nuevo)

def buscar_por_id(id_buscado):
    for id_nuevo in inventario:
        if id_buscado == id_nuevo.get("id"):
            print(f"\n{id_nuevo}")

#-------------------------------------------------------------------------------
salir = False
print(".-" * 10, "Sistema de inventario", ".-" * 10)
while salir == False:

    print("\n---Menu---")
    print("\t1.- Mostrar inventario")
    print("\t2.- Agregar nuevo producto")
    print("\t3.- Buscar producto por ID")
    print("\t4.- Salir")

    respuesta = int(input("Seleccione una opción: "))

    if respuesta == 1: 
        mostrar_inventario()
    elif respuesta == 2:
        agregar_producto()
    elif respuesta == 3:
        id_buscado = int(input("Coloque el id del producto que busca"))
        buscar_por_id(id_buscado)
    else:
        print("\nAdiós")
        salir = True
```
