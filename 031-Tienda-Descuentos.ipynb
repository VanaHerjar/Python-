```python
#Programa: Descuento por ser miembro y commprar +1k en productos 

print("|------------Sistema-de-descuentos------------|")

MONTO_MINIMO = 1000

monto_inicial = float(input("Cual fue el monto de su compra?"))

miembro = input("Eres miembro de mesa? (si/no)").lower().strip()



if (monto_inicial >= MONTO_MINIMO) and (miembro == "si"):
    
    descuento = monto_inicial * 0.1

    print("\n\nFelicidades, has obtenido un descuento del 10%")

elif miembro == "si":

    descuento = monto_inicial * 0.05

    print("\n\nFelicidades, has obtenido un descuento del 5%")

else:
    descuento = 0

monto_final = monto_inicial - descuento

print("Monto de la compra: ", monto_inicial)

print("Monto del descuento :", descuento)

print("Monto final de la compra con descuento :", monto_final)
```
