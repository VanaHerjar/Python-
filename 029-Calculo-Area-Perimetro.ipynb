```python
#Programa: Calcular el area y perimetro de un rectangulo

ladoa = int(input("Ingrese la base del rectangulo"))

ladob = int(input("Ingrese la altura largo del rectangulo"))


if(ladoa != ladob):
    area = (ladoa * ladob)

    perimetro = 2 * (ladoa + ladob)

    print("El area es :", area)

    print("El perimetro es :", perimetro)
else:
    print("Eso es un cuadrado")
```
