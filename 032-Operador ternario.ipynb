```python
#Programa: Fitness

META_PASOS_X_DIA = 10000  #10K
CALORIAS_X_PASO = 0.04

nombre = input("Ingresa tu nombre: ")

pasos_x_dia = int(input("Cuantos pasos a dado en el día: "))

calorias_quemadas = pasos_x_dia * CALORIAS_X_PASO

meta_alcanzada = "Si" if pasos_x_dia >= META_PASOS_X_DIA else "No"

print(f'Usted a quemado un total de: {calorias_quemadas} calorias')

print("Cumplio con la meta diaria: ", meta_alcanzada)
```

```python
#Programa: Reserva de hotel 

nombre = input("Ingrese su nombre")

dias_a_quedarse = int(input("Cuantos dias se quedara en el hotel"))

vista_al_mar = input("Desea un cuarto con vista al mar (Si/No)")

cuarto_vista_mar = 190.50 if (vista_al_mar.strip().lower() == "si") else 150.50

costo_total_a_pagar = cuarto_vista_mar * dias_a_quedarse

print("\n\n\n")

print("-.-" * 30)

print(f'Nombre del usuario: {nombre}')

print(f'Cuarto con vista al mar: {vista_al_mar}')

print(f'Días de estancia en el hotel: {dias_a_quedarse}')

print(f'Monto x día: {cuarto_vista_mar:.2f}')

print(f'Total a pagar: {costo_total_a_pagar:.2f}')
```

```python
#Programa: Comparar numeros

numero1 = int(input("Ingrese el primer numero: "))

numero2 = int(input("Ingrese el segundo numero: "))

numero_mayor = numero1 if numero1 > numero2 else numero2

print("\n")

print("-.-" * 14)

print(f'El numero mayor es: {numero_mayor}')
```

```python
#Programa: Cajero automatico 

Salir = False

Saldo = 1000.00

while not Salir:
    print("-." * 10, "Cajero automatico", "-." * 10)
    print("""1.- Consultar Saldo
2.- Retirar
3.- Depositar
4.- Salir
""")
    respuesta = int(input("Elija 1 opcion"))

    if respuesta == 1:
        print(f'Su saldo es : {Saldo:.2f} Soles\n')
    elif respuesta == 2:
        se_puede_retirar = int(input("Cuanto dinero necesitas retirar: "))
        if se_puede_retirar <= Saldo:
            Saldo -= se_puede_retirar
            print(f'Se ha retirado {se_puede_retirar} Soles')
            print(f'Saldo restante: {Saldo:.2f} soles\n')
        else:
            print("No cuenta con fondos suficientes\n")
    elif respuesta == 3:
        depositar = int(input("Cuanto dinero desea depositar: "))
        if depositar > 0:
            Saldo += depositar  
            print(f'Se ha depositado {depositar} soles')
            print(f'Saldo actual: {Saldo:.2f} soles\n')
        else:
            print("Error: El monto a depositar debe ser mayor a 0.\n")
    elif respuesta == 4: 
        print("Saliendo del sistema...")
        Salir = True
    else:
        print("No existe esa funcion\n")
    
```
