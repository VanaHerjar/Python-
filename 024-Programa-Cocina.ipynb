```python
#Se requiere hacer un programa para cocina 
#Datos necesarios:
#Nombre empleado
#Edad empleado
#Salario empleado
#Es jefe de departamento
#Usaremos el == para comparar, el cual nos dara el valor de True/False
```

```python
print("|---------------Programa: Cocina---------------|")

#------------Ingresamos datos basicos------------#

nombre = input("Ingrese su nombre")
edad = int(input("Ingrese su edad"))
salario = float(input("Ingrese su salario"))
es_jefe_departamento = input("Es jefe de departamento? (Si/No)")

#------------------True or false------------------#

es_jefe_departamento = es_jefe_departamento.lower() == 'si'

#----------------Valores a imprimir---------------#
print(" ")
print("--------------Datos del empleado--------------")
print("Su nombre es: ", nombre)
print("Su edad es: ", edad)
print(f"Su salario es: {salario:.2f}")
print("Es jefe de departamento: ", es_jefe_departamento)
print("------------------------------------------------")
```
